```python
import numpy as np

x_np = np.random.random((4,4))
print(x_np)

x_pt = torch.tensor(x_np)
print(x_pt)
```

Why numpy tensors are different from pytorch tensors ? Are their computing workflows different ?

```python
print(x_np.dtype, x_pt.dtype)

#Lets force them to be 32-bit floats
x_np = np.asarray(x_np, dtype=np.float32)
x_pt = torch.tensor(x_np, dtype=torch.float32)
print(x_np.dtype, x_pt.dtype) 
```

Let's say we wanted to find every value in a tensor that was greater than 0.5. Both PyTorch and NumPy would allow us to use the standard logic operators to check for things like this.

```python
b_np = (x_np > 0.5)
print(b_np)
print(b_np.dtype)

b_pt = (x_pt > 0.5)
print(b_pt)
print(b_pt.dtype)
```

```python
print(np.sum(x_np))


torch.sum(x_pt)
```

```python
np.transpose(x_np) 

torch.transpose(x_pt,0,1)
```
